```python
import pandas as pd
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    database="qversity",
    user="qversity-admin",
    password="qversity-admin",
    port="5432"
)

digital_engagement_df = pd.read_sql_query("SELECT * FROM public_silver.stg_digital_engagement", conn)

digital_engagement_df.head()
```

```python
preferred_channel_counts = (
    digital_engagement_df.groupby("preferred_channel")
      .agg(count=("preferred_channel", "size"))
      .reset_index()
      .sort_values(by="preferred_channel", ascending=False)
)
print(preferred_channel_counts.to_string())
```

```python
push_notifications_counts = (
    digital_engagement_df.groupby("push_notifications")
      .agg(count=("push_notifications", "size"))
      .reset_index()
      .sort_values(by="push_notifications", ascending=False)
)
push_notifications_counts.head()
```

```python
paperless_statements_counts = (
    digital_engagement_df.groupby("paperless_statements")
      .agg(count=("paperless_statements", "size"))
      .reset_index()
      .sort_values(by="paperless_statements", ascending=False)
)
paperless_statements_counts.head()
```

```python
import pandas as pd
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    database="qversity",
    user="qversity-admin",
    password="qversity-admin",
    port="5432"
)

customers_df = pd.read_sql_query("SELECT * FROM silver.stg_customer", conn)

customers_df.head()
```

Attributes count

```python
gender_counts = (
    customers_df.groupby("gender")
      .agg(count=("gender", "size"))
      .reset_index()
      .sort_values(by="count", ascending=False)
)
print(gender_counts)
```

```python
city_counts = (
    customers_df.groupby("city")
      .agg(count=("city", "size"))
      .reset_index()
      .sort_values(by="city", ascending=False)
)
print(city_counts.to_string())
```

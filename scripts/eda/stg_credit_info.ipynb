```python
import pandas as pd
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    database="qversity",
    user="qversity-admin",
    password="qversity-admin",
    port="5432"
)

credit_info_df = pd.read_sql_query("SELECT * FROM public_silver.stg_credit_info", conn)

credit_info_df.head()
```

```python
bankruptcy_flag_counts = (
    credit_info_df.groupby("bankruptcy_flag")
      .agg(count=("bankruptcy_flag", "size"))
      .reset_index()
      .sort_values(by="bankruptcy_flag", ascending=False)
)
print(bankruptcy_flag_counts.to_string())
```

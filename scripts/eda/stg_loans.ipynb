```python
import pandas as pd
import psycopg2
print("Connecting to the database...")
conn = psycopg2.connect(
    host="localhost",
    database="qversity",
    user="qversity-admin",
    password="qversity-admin",
    port="5432"
)
print("Connection established successfully!")
loans_df = pd.read_sql_query("SELECT * FROM silver.stg_loans", conn)
print("Data loaded successfully!")
loans_df.head()
```

```python
type_counts = (
    loans_df.groupby("type")
      .agg(count=("type", "size"))
      .reset_index()
      .sort_values(by="type", ascending=False)
)
print(type_counts.to_string())
```

```python
currency_counts = (
    loans_df.groupby("currency")
      .agg(count=("currency", "size"))
      .reset_index()
      .sort_values(by="currency", ascending=False)
)
print(currency_counts.to_string())
```

```python
status_counts = (
    loans_df.groupby("status")
      .agg(count=("status", "size"))
      .reset_index()
      .sort_values(by="status", ascending=False)
)
print(status_counts.to_string())
```

```python
collateral_type_counts = (
    loans_df.groupby("collateral_type")
      .agg(count=("collateral_type", "size"))
      .reset_index()
      .sort_values(by="collateral_type", ascending=False)
)
print(collateral_type_counts.to_string())
```

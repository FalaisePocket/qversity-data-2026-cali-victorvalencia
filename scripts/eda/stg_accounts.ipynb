```python
import pandas as pd
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    database="qversity",
    user="qversity-admin",
    password="qversity-admin",
    port="5432"
)

accounts_df = pd.read_sql_query("SELECT * FROM silver.stg_accounts", conn)

accounts_df.head()
```

```python
account_type_counts = (
    accounts_df.groupby("account_type")
      .agg(count=("account_type", "size"))
      .reset_index()
      .sort_values(by="account_type", ascending=False)
)
print(account_type_counts.to_string())
```

```python
status_counts = (
    accounts_df.groupby("status")
      .agg(count=("status", "size"))
      .reset_index()
      .sort_values(by="status", ascending=False)
)
print(status_counts.to_string())
```

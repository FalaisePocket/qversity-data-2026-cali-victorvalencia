```python
import pandas as pd
import psycopg2

conn = psycopg2.connect(
    host="localhost",
    database="qversity",
    user="qversity-admin",
    password="qversity-admin",
    port="5432"
)

accounts_df = pd.read_sql_query("SELECT * FROM silver.stg_transactions", conn)

accounts_df.head()
```

```python
type_counts = (
    accounts_df.groupby("type")
      .agg(count=("type", "size"))
      .reset_index()
      .sort_values(by="type", ascending=False)
)
print(type_counts.to_string())
```

```python
category_counts = (
    accounts_df.groupby("category")
      .agg(count=("category", "size"))
      .reset_index()
      .sort_values(by="category", ascending=False)
)
print(category_counts.to_string())
```

```python
merchant_counts = (
    accounts_df.groupby("merchant")
      .agg(count=("merchant", "size"))
      .reset_index()
      .sort_values(by="merchant", ascending=False)
)
print(merchant_counts.to_string())
```

```python
status_counts = (
    accounts_df.groupby("status")
      .agg(count=("status", "size"))
      .reset_index()
      .sort_values(by="status", ascending=False)
)
print(status_counts.to_string())
```
